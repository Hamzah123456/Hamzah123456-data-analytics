```python
import numpy as np
```

```python
 ## Looping Through Arrays
```

```python
arrA = np.array([[11, 12, 13], [14, 15, 16]])
print(arrA)
```

```python
for x in arrA:
    print(x * 2)
```

```python
for x in arrA:
    for y in x:
        print(y * 2)
```

```python
# The first loop goes over the rows of arrA, so each "x" is a whole row. And multiplying it by 2 gives back an entire array with each value doubled. The seccond loop goes one level lower. For "y in x" ut pulls out each individual number from the row, so it prints each element doubled one at a time instead of a row.
```

```python
for x in arrA:
    for y in x:
        loop_result = x * 2
print(loop_result)
# This prints the whole row doubled, not just one number. That's because we set loop_result to x*2 (the row) instead of y*2 (one element). Also, print is outside the loop, so it only shows the last value loop_result was set to.
```

```python
## Joining and Stacking Arrays
```

```python
joined_array_1 = np.concatenate((arrA, arrA), axis=0)
print(joined_array_1)
```

```python
joined_array_2 = np.concatenate((arrA, arrA), axis=1)
print(joined_array_2)
```

```python
stacked_array_1 = np.stack((joined_array_1, joined_array_1))
print(stacked_array_1)
```

```python
stacked_array_2 = np.stack((joined_array_1, joined_array_2))
print(stacked_array_2)
# Error: "all input arrays must have the same shape"
# joined_array_1 is shape (4, 3) and joined_array_2 is shape (2, 6).
# np.stack() requires arrays to be the exact same shape because it adds a new axis,
# so it can't combine arrays with different dimensions.
```

```python
another_joined_array = np.concatenate((joined_array_1, arrA))
print(another_joined_array)
```

```python
another_2 = np.concatenate((joined_array_1, arrA), axis=1)
print(another_2)
# This throws a shape mismatch error. When concatenating on axis=1, the arrays need to have
# the same number of rows. joined_array_1 has 4 rows but arrA only has 2 rows, so NumPy
# can't line them up side by side.
```

```python
h_stacked = np.hstack((arrA, arrA))
# hstack stacks arrays horizontally (side by side), joining them along columns.
print(h_stacked)

v_stacked = np.vstack((arrA, arrA))
# vstack stacks arrays vertically (on top of each other), joining them along rows.
print(v_stacked)

d_stacked = np.dstack((arrA, arrA))
# dstack stacks arrays depth-wise, adding a new third dimension to combine them.
print(d_stacked)
```

```python
## Splitting Arrays
```

```python
splitsky = np.array_split(joined_array_1, 2)
print(splitsky)
# Result: 2 arrays, each is 2-dimensional with shape (2, 3).
# joined_array_1 had 4 rows, so splitting into 2 gives two even chunks of 2 rows each.
```

```python
spliterino = np.array_split(joined_array_1, 3)
print(spliterino)
# Result: 3 arrays, each is 2-dimensional.
# Since 4 rows can't divide evenly into 3, NumPy makes the first array have 2 rows
# and the other two arrays have 1 row each. Shapes: (2, 3), (1, 3), (1, 3).
```

```python
## Searching and Replacing Array Values
```

```python
result_1 = np.where(arrA % 2 == 0, arrA, 'odd')
print(result_1)
```

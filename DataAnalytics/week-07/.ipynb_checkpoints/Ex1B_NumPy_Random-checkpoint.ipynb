```python
## Generating Random Choices from an Array
```

```python
students = np.array(['Hamzah', 'Jake', 'Carlos', 'Diana', 'Ethan', 'Fatima',
                     'Gabriel', 'Hana', 'Isaiah', 'Jasmine', 'Kevin', 'Layla'])
```

```python
call_on = np.random.choice(students)
print(call_on)
```

```python
breakout_groups = np.random.choice(students, size=(3, 4))
print(breakout_groups)
# This wouldn't be an effective way to assign breakout groups because np.random.choice
# samples with replacement by default, meaning the same student can appear multiple times
# while others might not appear at all. For real breakout groups, you'd want each student
# assigned to exactly one group, which would require sampling without replacement.
```

```python
call_on_weighted = np.random.choice(
    students,
    size=5,
    p=[0.30, 0.15, 0.10, 0.10, 0.10, 0.05, 0.05, 0.05, 0.05, 0.05, 0.00, 0.00]
)
print(call_on_weighted)
# p values explained: Alex has a 30% chance (highest maybe they've been quiet lately
# and need to participate more). Brianna has 15%, Carlos through Ethan each have 10%,
# Fatima through Jasmine each have 5%, and Kevin and Layla have 0% (maybe they
# already presented today and don't need to be called on again).
# All p values must add up to 1.0.
```

```python
## Generating a Normal Distribution
```

```python
bell_curve = np.random.normal(loc=10, scale=5, size=(4, 5))
print(bell_curve)
```
